# Câu 2: Thống kê dữ liệu khuyết thiếu

```python
print("Họ và tên: Lê Nguyễn Huy Phát - MSSV: 6451071057")
print("-" * 30)

# Lấy ra 5 thuộc tính có số lượng giá trị rỗng nhiều nhất
total = df.isnull().sum().sort_values(ascending=False)
percent_1 = df.isnull().sum() / df.isnull().count() * 100
percent_2 = (round(percent_1, 1)).sort_values(ascending=False)
missing_data = pd.concat([total, percent_2], axis=1, keys=['Total', '%'])

# Hiển thị 5 dòng đầu tiên (5 thuộc tính thiếu nhiều nhất)
print("5 thuộc tính có số lượng giá trị rỗng nhiều nhất:")
display(missing_data.head(5))
```

# Câu 3: Trực quan hóa dữ liệu theo Độ tuổi, Giới tính và Sống sót

```python
print("Họ và tên: Lê Nguyễn Huy Phát - MSSV: 6451071057")
print("-" * 30)
# Vẽ biểu đồ thống kê giữa Độ tuổi, Giới tính và Khả năng sống sót
survived = 'survived'
not_survived = 'not survived'
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))

women = df[df['sex'] == 'female']
men = df[df['sex'] == 'male']

# Biểu đồ cho Nữ (Female)
ax = sns.histplot(women[(women['survived'] == 1)].age.dropna(), bins=18, label=survived, ax=axes[0], kde=False, color='g')
ax = sns.histplot(women[(women['survived'] == 0)].age.dropna(), bins=40, label=not_survived, ax=axes[0], kde=False, color='y')
ax.legend()
ax.set_title('Female')

# Biểu đồ cho Nam (Male)
ax = sns.histplot(men[(men['survived'] == 1)].age.dropna(), bins=18, label=survived, ax=axes[1], kde=False, color='g')
ax = sns.histplot(men[(men['survived'] == 0)].age.dropna(), bins=40, label=not_survived, ax=axes[1], kde=False, color='y')
ax.legend()
ax.set_title('Male')

plt.tight_layout()
plt.show()
```

### 5.1 Thống kê số người đi một mình và đi cùng người thân

```python
print("Họ và tên: Lê Nguyễn Huy Phát - MSSV: 6451071057")
print("-" * 30)

# Thống kê số người đi cùng
for dataset in [df]:
    dataset['relatives'] = dataset['sibsp'] + dataset['parch']
    dataset.loc[dataset['relatives'] > 0, 'not_alone'] = 0
    dataset.loc[dataset['relatives'] == 0, 'not_alone'] = 1
    dataset['not_alone'] = dataset['not_alone'].astype(int)

# Hiển thị số lượng giá trị
print(df['not_alone'].value_counts())
```

# Câu 6: Tiền xử lý dữ liệu

### 6.1 Xử lý thuộc tính cabin

```python
print("Họ và tên: Lê Nguyễn Huy Phát - MSSV: 6451071057")
print("-" * 30)
import re

# Khai báo từ điển ánh xạ ký tự boong sang số
deck = {"A": 1, "B": 2, "C": 3, "D": 4, "E": 5, "F": 6, "G": 7, "U": 8}

for dataset in [df]:
    dataset['cabin'] = dataset['cabin'].fillna("U0")
    dataset['deck'] = dataset['cabin'].map(lambda x: re.compile("([a-zA-Z]+)").search(x).group())
    dataset['deck'] = dataset['deck'].map(deck)
    dataset['deck'] = dataset['deck'].fillna(0)
    dataset['deck'] = dataset['deck'].astype(int)

# Xóa cột cabin sau khi đã trích xuất xong
df = df.drop(['cabin'], axis=1)
```

### 6.2 Xem xét và xóa các thuộc tính không cần thiết

```python
print("Họ và tên: Lê Nguyễn Huy Phát - MSSV: 6451071057")
print("-" * 30)
# Kiểm tra mô tả thuộc tính ticket
print(df['ticket'].describe())
print("-" * 30)

# Thực hiện xóa các thuộc tính không cần thiết
df = df.drop(['ticket'], axis=1)
df = df.drop(['boat'], axis=1)
df = df.drop(['body'], axis=1)
df = df.drop(['home.dest'], axis=1)

print("Đã xóa các thuộc tính không cần thiết thành công!")
```

### 6.3 Điền giá trị còn thiếu vào thuộc tính độ tuổi (age)

```python
print("Họ và tên: Lê Nguyễn Huy Phát - MSSV: 6451071057")
print("-" * 30)
# Điền giá trị còn thiếu cho cột age bằng số liệu ngẫu nhiên dựa trên mean và std
for dataset in [df]:
    mean = df["age"].mean()
    std = df["age"].std()
    is_null = dataset["age"].isnull().sum()

    # Tạo giá trị ngẫu nhiên giữa mean - std và mean + std
    rand_age = np.random.randint(mean - std, mean + std, size=is_null)

    # Điền vào các vị trí NaN
    age_slice = dataset["age"].copy()
    age_slice[np.isnan(age_slice)] = rand_age
    dataset["age"] = age_slice
    dataset["age"] = df["age"].astype(int)

# Kiểm tra lại số lượng giá trị null của age (phải bằng 0)
print("Số lượng giá trị null còn lại của age:", df["age"].isnull().sum())
```

### 6.4 Điền giá trị còn thiếu vào thuộc tính Cổng lên tàu (embarked)

```python
print("Họ và tên: Lê Nguyễn Huy Phát - MSSV: 6451071057")
# Xem thông tin thống kê của embarked
print(df['embarked'].describe())
print("-" * 30)

# Điền giá trị phổ biến nhất ('S') vào các ô còn thiếu
common_value = 'S'
for dataset in [df]:
    dataset['embarked'] = dataset['embarked'].fillna(common_value)

print("Đã điền xong giá trị thiếu cho embarked!")
```
